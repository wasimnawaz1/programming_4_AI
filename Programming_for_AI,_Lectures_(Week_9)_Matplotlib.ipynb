### Line Plot

**Line plots** are predominantly useful for conveying changes over space or time. They are mostly used to plot data along a scale divided into equal intervals, e.g., time.

- To generate a line plot: use `plt.plot(x, y, label=...)`
- Add titles and labels using `plt.title()`, `plt.xlabel()`, `plt.ylabel()`
- Add a legend using `plt.legend()`
- Multiple plots can be drawn on the same figure

```python
import matplotlib.pyplot as plt

x1 = [1,2,3,4]
y1 = [2,4,3,5]
plt.plot(x1, y1, label = "line 1")
plt.legend()
plt.show()
```

```python
x1 = [1,2,3]
y1 = [2,4,1]
plt.plot(x1, y1, label = "line 1")
# naming the x axis
plt.xlabel('x - axis')
# naming the y axis
plt.ylabel('y - axis')
# giving a title to my graph
plt.title('My first line plot')
plt.legend()
plt.show()
```

```python
# Generating x and y points to plot
import numpy as np
x1= np.arange(0, 10*(np.pi), 0.1)
y1 = np.cos(x1) # cosine function from Numpy library


# potting the points
plt.plot(x1, y1, label = 'cosine')
plt.xlabel('Angle') # naming the x axis
plt.ylabel('Amplitude') # naming the y axis

# Generating x and y points for the second plot
x2= np.arange(0, 10*(np.pi), 0.1)
y2 = np.cos(x2)*0.1*np.arange(10*(np.pi),0,-0.1)
##       cosine*0.1* decreasing function

# decaying cosine function
# potting the points
plt.plot(x2, y2, label = 'Decaying cosine')
plt.legend() # show a legend on the plot
plt.title('Two functions on the same graph') # gives a title to the plot
plt.show() # shows the graph, and removes the text output.
```

### Scatter Plot

A **scatter plot** is used to visualize any two variables from a dataset in two dimensions. It uses dots or marks to plot values of two variables.

Scatter plots help us observe:
- **Positive linear relationship**: increase in one variable causes increase in the other
- **Negative linear relationship**: increase in one variable causes decrease in the other

If the dots are **color-coded**, additional variables can be shown in two dimensions.

Use `plt.rcParams['figure.figsize']` to control the plot size for better readability.

Matplotlib's `rcParams` (runtime configuration parameters) are dictionary-like settings that control default visual styles (colors, fonts, sizes) for plots, allowing for consistent styling without repeating code.

```python
# Adjusting the figure size using rcParams function
plt.rcParams['figure.figsize'] = [8,6]
# Scatter plot
plt.scatter(housing_price['Avg. Area Income'].iloc[::10],
            housing_price['Price'].iloc[::10], color='red', marker='+')
# Giving title and label names
plt.title('Avg. Area Income Vs Price', fontsize=14)
plt.xlabel('Avg. Area Income', fontsize=14)
plt.ylabel('Price', fontsize=14)
plt.grid(True)
plt.show()
```

### Bar Chart

**Bar charts** are used when we have **categorical or discrete data** that can take one of a small set of values.

- The lengths of rectangular bars are proportional to the values they represent
- Suitable for visualizing species counts, categorical distributions, etc.
- We use the **Iris dataset** here, which contains three iris flower species.
- `bar` plots vertical bars, whereas `barh` plots horizontal bars.
The Iris dataset can be downloaded from Kaggle or from `sklearn` library.

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()   # 3 target variables / categories / classes of iris flowers
iris_data = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_data['species'] = iris.target  # Adds the target labels
iris_data.info()
iris_data
```

```python
# Generating a bar chart
plt.barh(iris_data['species'], iris_data['sepal width (cm)'])

# Giving title and label names
plt.title('Sepal Length (cm) Vs Species', fontsize=14)
plt.xlabel('Species', fontsize=14)
plt.ylabel('Sepal Length (cm)', fontsize=14)
plt.show()
```

```python
# Generating a bar chart
plt.bar(iris_data['species'], iris_data['sepal length (cm)'], label = 'Sepal Length (cm)')
plt.bar(iris_data['species'], iris_data['sepal width (cm)'], label = 'Sepal Width (cm)', color='green')
# Giving title and label names
plt.title('Sepal Length (cm) and With (cm) Vs Species', fontsize=14)
plt.xlabel('Species', fontsize=14)
plt.ylabel('Sepal Length (cm) and Width (cm)', fontsize=14)
plt.legend()
plt.show()
```

### Checking Assumption: Independence of Features

We can check if features are **independent of each other** using a **scatter matrix**:

- **Diagonal entries**: histograms (or KDE plots) of each individual feature
- **Off-diagonal entries**: scatter plots of every pair of features

Interpretation:
- Points forming a **circle** → features are **independent**
- Points forming an **angled oval** → features are **dependent**

The `diagonal` option can be set to `'kde'` (kernel density estimation) or `'hist'`.

### Selecting Important Features / Variables

**Feature selection** picks features that contribute most to the target variable. It helps:
- Reduce **computational load**
- Achieve **better accuracy**
- Remove irrelevant or less important features

### Correlation-Based Feature Selection

**Correlation** is a statistical term that measures the dependency between two features as a number ranging from **–1 to 1**:

- **Positive correlation**: both variables increase/decrease together
- **Negative correlation**: one increases while the other decreases
- **Near zero**: features are nearly independent

We use:
- `corr()` method to compute the correlation matrix
- `sns.heatmap()` from Seaborn to visualize it

Features with a **small correlation** to the target variable can be dropped.

```python
### Check the correlation of each feature with target variable `price`
### Features with a **small correlation** to the target variable can be dropped
### (e.g., `Avg. Area Number of Bedrooms` has correlation 0.17 with `Price`).


import seaborn as sns
# Drop the 'Address' column before calculating correlation as it is a string column.
housing_price_numeric = housing_price.drop(columns=['Address'])
corrmat = housing_price_numeric.corr()
feature_ind = corrmat.index
plt.figure(figsize=(8, 8))
#Plotting correlation matrix as an sns heat map
sns.heatmap(housing_price_numeric[feature_ind].corr(), annot=True, cmap="RdYlGn")
plt.show()
```

```python
import pandas as pd

# Create a sample DataFrame with string columns
data = {
    'String_Numbers': ['10', '20', '30', '40', '50'],
    'String_Mixed': ['100', '200', 'not_a_number', '400', '500'],
    'Other_String': ['apple', 'banana', 'orange', 'grape', 'kiwi']
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("\nData types before conversion:")
print(df.dtypes)

# Convert 'String_Numbers' column to numeric
df['Numeric_Numbers'] = pd.to_numeric(df['String_Numbers'])

# Convert 'String_Mixed' column to numeric
df['Numeric_Mixed'] = pd.to_numeric(df['String_Mixed'], errors='coerce')         #, errors='coerce' forces errors to NaN

print("\nDataFrame after conversions:")
print(df)
print("\nData types after conversion:")
print(df.dtypes)

# Note: Columns like 'Other_String' containing non-numeric words cannot be
# directly converted to numeric.
```

```python
# One-hot encoding creates new binary columns for each unique category.
# This is suitable for models that require numerical input.

import pandas as pd

# Create a sample DataFrame with a categorical string column
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Laptop', 'Mouse'],
    'Price': [1200, 25, 75, 1300, 30],
    'Color': ['Silver', 'Black', 'Black', 'Grey', 'Silver']
}
df_categorical = pd.DataFrame(data)

print("Original DataFrame with categorical data:")
print(df_categorical)
print("\nData types before one-hot encoding:")
print(df_categorical.dtypes)

# Perform one-hot encoding on the 'Product' and 'Color' columns
df_encoded = pd.get_dummies(df_categorical, columns=['Product', 'Color'], dtype=int)

print("\nDataFrame after one-hot encoding:")
print(df_encoded)
print("\nData types after one-hot encoding:")
print(df_encoded.dtypes)

```

```python
# Note: Label Encoding is suitable for ordinal categorical data (where there's an inherent order)

import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Re-using the sample DataFrame with categorical string columns
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Laptop', 'Mouse'],
    'Price': [1200, 25, 75, 1300, 30],
    'Color': ['Silver', 'Black', 'Black', 'Grey', 'Silver']
}
df_categorical = pd.DataFrame(data)

print("Original DataFrame:")
print(df_categorical)

# Initialize the LabelEncoder
label_encoder = LabelEncoder()

# Apply LabelEncoder to the 'Product' column
df_categorical['Product_Encoded'] = label_encoder.fit_transform(df_categorical['Product'])
print(f"\nMapping for Product: {list(label_encoder.classes_)} -> {list(range(len(label_encoder.classes_)))}")

# Apply LabelEncoder to the 'Color' column
df_categorical['Color_Encoded'] = label_encoder.fit_transform(df_categorical['Color'])
print(f"Mapping for Color: {list(label_encoder.classes_)} -> {list(range(len(label_encoder.classes_)))}")

print("\nDataFrame after Label Encoding:")
print(df_categorical)
print("\nData types after encoding:")
print(df_categorical.dtypes)

```

```python
### Homework: Labeling Boxplot

### Use ? with a method to read its documentation, for example, plt.boxplot?

import matplotlib.pyplot as plt
import numpy as np

data = [0, 1, 10, 15, 4, -6, -15 -2, 30, 40, -20, 11]

plt.figure(figsize=(8, 6))
box_plot_dict = plt.boxplot(data, vert=True, patch_artist=True)   # horizontal / vertical orientation and filling of the box

plt.title('Box Plot of Sample Data with Quartile Labels')
plt.ylabel('Values')
plt.grid(True)

# Extracting and labeling quartiles and median
for i, line in enumerate(box_plot_dict['medians']):
    x, y = line.get_xydata()[0]
    median = line.get_ydata()[0]
    plt.text(x + 0.1, median, f'Median: {median:.2f}', va='center', color='black', fontsize=9)

for i, box in enumerate(box_plot_dict['boxes']):
    # Box contains 25th and 75th percentile (Q1 and Q3)
    # Accessing the y-coordinates from the PathPatch vertices
    y_coords = box.get_path().vertices[:, 1]
    q1 = np.min(y_coords) # Q1 is the minimum y-coordinate of the box
    q3 = np.max(y_coords) # Q3 is the maximum y-coordinate of the box

    # Get x-coordinate for text placement from the box's vertices
    x_coords = box.get_path().vertices[:, 0]
    x_position = np.min(x_coords) # Use the left-most x-coordinate of the box

    # Place labels slightly above/below the box edges
    plt.text(x_position - 0.1, q1, f'Q1: {q1:.2f}', va='center', ha='right', color='blue', fontsize=9)
    plt.text(x_position - 0.1, q3, f'Q3: {q3:.2f}', va='center', ha='right', color='blue', fontsize=9)

plt.tight_layout()
plt.show()
```
